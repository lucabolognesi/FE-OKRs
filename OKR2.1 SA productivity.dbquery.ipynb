```sql
-- Use cases for Solutions Architects in Italy Strategic Core (stages 2 & 3)
-- Excluding: Gruppo Hera, ALIA SERVIZI AMBIENTALI SPA, Snam Spa
WITH filtered_use_cases AS (
  SELECT
    solution_architect_user_name AS solutions_architect,
    usecase_id,
    usecase_name,
    account_name,
    stage_number,
    stage_name_ui,
    target_onboarding_fiscal_year_quarter,
    estimated_quarterly_dollar_dbus
  FROM main.gtm_gold.rpt_use_case_detail
  WHERE sales_subregion_level_3 = 'Italy Strategic Core'
    AND account_name NOT IN ('Gruppo Hera', 'ALIA SERVIZI AMBIENTALI SPA', 'Snam Spa')
    AND stage_number IN (2, 3)
    AND solution_architect_user_name IS NOT NULL
)

SELECT
  solutions_architect,

  -- FY27 Q2 (current quarter)
  SUM(CASE WHEN target_onboarding_fiscal_year_quarter = "FY'27 Q2" THEN 1 END) AS uco_count_fy27q2,
  ROUND(SUM(CASE WHEN target_onboarding_fiscal_year_quarter = "FY'27 Q2" THEN estimated_quarterly_dollar_dbus ELSE 0 END), 2) AS total_quarterly_dbus_fy27q2,
  ROUND(TRY_DIVIDE(
    SUM(CASE WHEN target_onboarding_fiscal_year_quarter = "FY'27 Q2" THEN estimated_quarterly_dollar_dbus ELSE 0 END),
    NULLIF(SUM(CASE WHEN target_onboarding_fiscal_year_quarter = "FY'27 Q2" THEN 1 END), 0)
  ), 2) AS avg_quarterly_dbus_per_sa_fy27q2,

  -- FY27 Q3 (next quarter)
  SUM(CASE WHEN target_onboarding_fiscal_year_quarter = "FY'27 Q3" THEN 1 END) AS uco_count_fy27q3,
  ROUND(SUM(CASE WHEN target_onboarding_fiscal_year_quarter = "FY'27 Q3" THEN estimated_quarterly_dollar_dbus ELSE 0 END), 2) AS total_quarterly_dbus_fy27q3,
  ROUND(TRY_DIVIDE(
    SUM(CASE WHEN target_onboarding_fiscal_year_quarter = "FY'27 Q3" THEN estimated_quarterly_dollar_dbus ELSE 0 END),
    NULLIF(SUM(CASE WHEN target_onboarding_fiscal_year_quarter = "FY'27 Q3" THEN 1 END), 0)
  ), 2) AS avg_quarterly_dbus_per_sa_fy27q3,

  -- Full FY27 (all quarters)
  SUM(CASE WHEN target_onboarding_fiscal_year_quarter LIKE "FY'27%" THEN 1 END) AS uco_count_fy27_all,
  ROUND(SUM(CASE WHEN target_onboarding_fiscal_year_quarter LIKE "FY'27%" THEN estimated_quarterly_dollar_dbus ELSE 0 END), 2) AS total_quarterly_dbus_fy27_all,
  ROUND(TRY_DIVIDE(
    SUM(CASE WHEN target_onboarding_fiscal_year_quarter LIKE "FY'27%" THEN estimated_quarterly_dollar_dbus ELSE 0 END),
    NULLIF(SUM(CASE WHEN target_onboarding_fiscal_year_quarter LIKE "FY'27%" THEN 1 END), 0)
  ), 2) AS avg_quarterly_dbus_per_sa_fy27_all,

  -- No date restriction
  COUNT(*) AS uco_count_all,
  ROUND(SUM(estimated_quarterly_dollar_dbus), 2) AS total_quarterly_dbus_all,
  ROUND(TRY_DIVIDE(SUM(estimated_quarterly_dollar_dbus), COUNT(*)), 2) AS avg_quarterly_dbus_per_sa_all

FROM filtered_use_cases
GROUP BY solutions_architect
ORDER BY solutions_architect
```
