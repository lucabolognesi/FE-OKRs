```sql
SELECT
  t1.deployable_account_name,
  t1.usecase_name,  
  t1.estimated_quarterly_dollar_dbus,
  t1.target_onboarding_fiscal_year_quarter,
  t1.target_live_fiscal_year_quarter,
  t1.stage_number,
  t1.has_implementation_partner,
  t1.implementation_partner_name,
  c.Name AS partner_tech_lead_name,
  t1.partner_summary[0].partner_account_manager AS partner_account_manager
FROM
  main.gtm_gold.rpt_use_case_detail t1
LEFT JOIN (SELECT Id, INITCAP(Name) AS Name FROM main.sfdc_bronze.contact GROUP BY Id, INITCAP(Name)) c
  ON t1.implementation_partners[0].implementation_partner_tech_lead = c.Id
WHERE
  t1.sales_subregion_level_3 = 'Italy Strategic Core'
  AND t1.account_name NOT IN ('Gruppo Hera', 'ALIA SERVIZI AMBIENTALI SPA', 'Snam Spa') --Exclude interim accounts.
  AND target_live_fiscal_year_quarter LIKE "FY'27%"
  AND stage_number BETWEEN 2 AND 5
ORDER BY t1.estimated_quarterly_dollar_dbus DESC
LIMIT 40 --TOP 20 UCOs by $dbus
```
