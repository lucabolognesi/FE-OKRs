```sql
SELECT
  --t1.deployable_account_name,
  COUNT(*) AS uco_count_total,
  COUNT(CASE WHEN t1.has_implementation_partner = true THEN 1 END) AS uco_count_with_partner,
  ROUND(100.0 * COUNT(CASE WHEN t1.has_implementation_partner = true THEN 1 END) / COUNT(*), 1) AS partner_attach_rate_pct
FROM
  main.gtm_gold.rpt_use_case_detail t1
WHERE
  t1.sales_subregion_level_3 = 'Italy Strategic Core'
  AND t1.account_name NOT IN ('Gruppo Hera', 'ALIA SERVIZI AMBIENTALI SPA', 'Snam Spa') --Exclude interim accounts.
  AND target_live_fiscal_year_quarter LIKE "FY'26%"
  AND stage_number BETWEEN 1 AND 7 -- includes live and lost.
GROUP BY ALL
ORDER BY uco_count_total DESC
```
