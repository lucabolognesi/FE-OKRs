```sql
-- Median days in Confirming (stage 4) by Solutions Architect - Italy Strategic Core
SELECT
  -- solution_architect_user_name as solution_architect,
  -- usecase_name,
  -- stage_number,
  -- days_in_stage,
  -- days_in_validating,
  -- days_in_scoping,
  -- days_in_evaluatig,
  -- days_in_confirming_all_time,
  -- days_in_confirming,
  --solution_architect_user_name AS solutions_architect,
  COUNT(*) AS uco_count_all,
  COUNT(CASE WHEN days_in_confirming_all_time = 0 THEN 1 END) AS uco_count_confirm_skipped,
  ROUND(PERCENTILE_CONT(0.5) WITHIN GROUP (ORDER BY days_in_confirming) FILTER (WHERE days_in_confirming > 0), 0) AS median_days_in_confirming,
  ROUND(AVG(CASE WHEN days_in_confirming > 0 THEN days_in_confirming END), 0) AS avg_days_in_confirming,
  MIN(CASE WHEN days_in_confirming > 0 THEN days_in_confirming END) AS min_days_in_confirming,
  MAX(CASE WHEN days_in_confirming > 0 THEN days_in_confirming END) AS max_days_in_confirming,
  ROUND(
    100.0 * COUNT(CASE WHEN days_in_confirming > 0 AND days_in_confirming <= 60 THEN 1 END)
    / NULLIF(COUNT(CASE WHEN days_in_confirming > 0 THEN 1 END), 0)
  , 1) AS pct_live_within_60_days
FROM main.gtm_gold.rpt_use_case_detail
WHERE sales_subregion_level_3 = 'Italy Strategic Core'
  AND target_live_fiscal_year_quarter LIKE "FY'27%"
  AND stage_number = 6 --UCOs gone live
  AND solution_architect_user_name IS NOT NULL
  AND deployable_account_name NOT IN ('Gruppo Hera', 'ALIA SERVIZI AMBIENTALI SPA', 'Snam Spa') --interim accounts
GROUP BY ALL
```
